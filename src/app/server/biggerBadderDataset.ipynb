```python
import pandas as pd
```

```python
df = pd.read_csv('megaGymDataset.csv')
```

```python
df = df.drop(['Desc', 'Equipment', 'RatingDesc', 'Rating', 'Unnamed: 0'], axis=1, errors='ignore')
```

```python
df.columns = ['exercise_type', 'exercise_category', 'target_muscle_group', 'Difficulty']
bigDSCols = df.columns.tolist()
bigDSCols
```

```python
smallDS = pd.read_csv('bulk.csv')
smallDSCols = smallDS.columns.tolist()

extraCols = [col for col in smallDSCols if col not in bigDSCols]
print(extraCols)
```

```python
for col in extraCols:
    df[col] = None
    
bigDSCols = df.columns.tolist()
bigDSCols
```

```python
df.to_csv('bigDataset.csv', index=False)
```

```python
import os

# Create the output directory if it doesn't exist
output_dir = 'megaDatasetBroken'
os.makedirs(output_dir, exist_ok=True)

# Split df into 10 chunks of 300 rows each
for i in range(10):
    start = i * 300
    end = start + 300
    chunk = df.iloc[start:end]
    chunk.to_csv(f'{output_dir}/megaGymDataset_part_{i+1}.csv', index=False)
```
